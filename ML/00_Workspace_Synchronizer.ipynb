## 4. Synchronization Engine
### 4.1 Compute Node Definition

```python
def execute_workspace_sync(source: str, target: str) -> None:
    """Melakukan sinkronisasi direktori secara inkremental."""
    if not os.path.exists(source):
        print(f"[SYNC] status=FAILED | error='source_not_found' | path='{source}'")
        return

    os.makedirs(target, exist_ok=True)
    # Menggunakan cp -r direktori/. untuk mengkopi folder tanpa membawa metadata akun asli
    sync_cmd: str = f"cp -r '{source}/.' '{target}/'"
    
    try:
        import subprocess
        subprocess.run(sync_cmd, shell=True, check=True, text=True, capture_output=True)
        print("[SYNC] status=COMPLETED")
    except subprocess.CalledProcessError as e:
        print(f"[SYNC] status=FAILED | error='{e.stderr.strip()}'")
```

### 4.2 Incremental Execution Loop

```python
execute_workspace_sync(source=SOURCE_DIR, target=TARGET_DIR)
```
