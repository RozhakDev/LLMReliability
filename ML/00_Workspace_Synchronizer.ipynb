4. (Opsional) Hapus *shortcut* sisa lainnya agar rapi.

```python
# 1. Authenticate & Mount New Drive
from google.colab import drive
drive.mount('/content/drive', force_remount=True)
```

```python
# 2. Final Rsync Engine
import os
import subprocess

DRIVE_ROOT = '/content/drive/MyDrive'
TARGET_DIR = os.path.join(DRIVE_ROOT, 'LLMReliability')
SOURCE_DIR = os.path.join(DRIVE_ROOT, 'SYNC_SOURCE')

print("[INFO] Memulai sinkronisasi...")

if not os.path.exists(SOURCE_DIR):
    print("\n❌ [FATAL ERROR] Shortcut bernama 'SYNC_SOURCE' tidak ditemukan!")
    print("Silakan ke Google Drive Anda, cari shortcut LLMReliability, lalu klik kanan -> Rename -> ketik 'SYNC_SOURCE'.")
else:
    # Buat target jika belum ada
    os.makedirs(TARGET_DIR, exist_ok=True)
    
    print(f"[INFO] Ditemukan sumber: {SOURCE_DIR}")
    print(f"[INFO] Menyiapkan target: {TARGET_DIR}")
    print(f"\n[INFO] Mulai Server-Side Sync. Ini akan mengkloning Qwen3 dan dataset...")
    
    command = f"rsync -avh --update --progress '{SOURCE_DIR}/' '{TARGET_DIR}/'"
    
    try:
        process = subprocess.run(command, shell=True, check=True, text=True, capture_output=True)
        print(process.stdout)
        print("\n✅ [SUCCESS] Workspace Anda tersinkronisasi sempurna!")
        print("Silakan langsung meluncur ke Notebook 03_Pilot_Experiment!")
    except subprocess.CalledProcessError as e:
        print("\n❌ [ERROR] Sinkronisasi gagal:")
        print(e.stderr)
```
