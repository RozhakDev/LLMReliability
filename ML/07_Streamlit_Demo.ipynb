### 2.2 Module Imports

```python
import json
import os
import subprocess
import time
from pyngrok import ngrok
from typing import Optional
```

### 2.3 Environment Variables

```python
PROJECT_ROOT: str = '/content/drive/MyDrive/LLMReliability'
SECRETS_PATH: str = '/content/drive/MyDrive/secrets.json'
NGROK_AUTH_TOKEN: Optional[str] = None

if os.path.exists(SECRETS_PATH):
    with open(SECRETS_PATH, 'r') as f:
        secrets = json.load(f)
    NGROK_AUTH_TOKEN = secrets.get("NGROK_TOKEN")

if not NGROK_AUTH_TOKEN:
    print("[ENVIRONMENT] status=ERROR | component='NGROK_TOKEN' | error='not found in secrets.json'")
else:
    ngrok.set_auth_token(NGROK_AUTH_TOKEN)
    print("[ENVIRONMENT] status=AUTHENTICATED | component='Ngrok'")
```

## 4. Application Server Deployment
### 4.1 Background Worker Process

```python
print("[DEPLOYMENT] component='Streamlit' | port=8501 | status=STARTING")
process = subprocess.Popen(
    ["python3", "-m", "streamlit", "run", "UI/app.py", "--server.address", "127.0.0.1", "--server.port", "8501"],
    cwd=PROJECT_ROOT,
    stdout=subprocess.PIPE, 
    stderr=subprocess.PIPE
)

time.sleep(5)
print("[DEPLOYMENT] component='Streamlit' | status=READY")
```

### 4.2 Network Tunnel Activation (Ngrok)

```python
try:
    print("[DEPLOYMENT] component='Ngrok' | status=CONNECTING")
    public_url = ngrok.connect(8501)
    print(
        f"[DEPLOYMENT] component='Ngrok' | "
        f"url='{public_url.public_url}' | "
        f"status=LIVE"
    )
except Exception as e:
    print(f"[DEPLOYMENT] component='Ngrok' | status=FAILED | error='{e}'")
```
